```python
import pandas as pd
import numpy as np
```

```python
capital_inicial=200_000
```

```python
port_inicial=pd.DataFrame(columns=['ACWI','IUIT','SPLG'])
port_inicial.loc['Peso']=[.30,.30,.40]
port_inicial.loc['Precios']=[91.25,20.21,52.14]
port_inicial.loc['No Acciones']=np.floor(capital_inicial*port_inicial.loc['Peso']/port_inicial.loc['Precios'])
port_inicial.loc['Monto Invertido']=port_inicial.loc['No Acciones']*port_inicial.loc['Precios']
port_inicial.loc['w_Real']=port_inicial.loc['Monto Invertido']/capital_inicial
port_inicial
```

```python
port_6m = pd.DataFrame(columns=['ACWI','IUIT','SPLG'])
port_6m.loc['No Acciones Inicial'] = port_inicial.loc['No Acciones']
port_6m.loc['Precio Actual'] = [87.45, 20.17, 77.15]
port_6m.loc['Valor Posición'] = (port_6m.loc['No Acciones Inicial'] *port_6m.loc['Precio Actual'])
port_6m.loc['Pesos Actuales'] = (port_6m.loc['Valor Posición'] /port_6m.loc['Valor Posición'].sum())
port_6m.loc['Pesos Objetivos'] = [.30, .30, .40]

port_6m
```
