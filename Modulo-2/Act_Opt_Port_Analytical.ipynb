*Hint:* Dado que tendrás dos restricciones de igualdad, es recomendable transformar el problema a su forma matricial, esto lo deberás de realizar después de derivar el lagrangiano e igualar las restricciones a cero. El resultado correcto es el siguiente.

$$w^*_{frontier} = \Sigma^{-1} R [R^T \Sigma^{-1} R]^{-1} P$$

donde $R_{nx2}=\begin{bmatrix} \mu & \overrightarrow{1}\end{bmatrix}$ y $P_{2x1}=\begin{bmatrix} \mu_p \\ 1 \end{bmatrix}$ y $\Sigma^{-1}_{nxn}$ es la inversa de la matriz cuadrada de varianza-covarianza


# 1. FRONTERA EFICIENTE DE PORTAFOLIOS

Se quiere minimizar la varianza del portafolio

$$
\min_{w} \sigma_p^2=w^T\Sigma w
$$

Sujeto a

$$
w^T\vec{1}=1
$$

$$
w^T\mu=\mu_p
$$

Utilizamos Lagrange

$$
L=w^T\Sigma w+\lambda(w^T\vec{1}-1)+\gamma(w^T\mu-\mu_p)
$$

Sacamos las derivadas

$$
\nabla L=
\begin{bmatrix}
\frac{\partial L}{\partial w}\\
\frac{\partial L}{\partial \lambda}\\
\frac{\partial L}{\partial \gamma}
\end{bmatrix}
$$

$$
\nabla L=
\begin{bmatrix}
2\Sigma w+\lambda\vec{1}+\gamma\mu\\
w^T\vec{1}-1\\
w^T\mu-\mu_p
\end{bmatrix}
$$

Igualamos a 0

$$
2\Sigma w+\lambda\vec{1}+\gamma\mu=0
$$

$$
w^T\vec{1}-1=0
$$

$$
w^T\mu-\mu_p=0
$$

Entonces las restricciones quedan

$$
w^T\vec{1}=1
$$

$$
w^T\mu=\mu_p
$$

Como tenemos dos restricciones las juntamos de forma matricial

$$
R=
\begin{bmatrix}
\mu & \vec{1}
\end{bmatrix}
$$

$$
P=
\begin{bmatrix}
\mu_p\\
1
\end{bmatrix}
$$

Entonces

$$
R^T=
\begin{bmatrix}
\mu^T\\
\vec{1}^T
\end{bmatrix}
$$

$$
R^Tw=
\begin{bmatrix}
\mu^Tw\\
\vec{1}^Tw
\end{bmatrix}
$$

Usando las restricciones

$$
R^Tw=
\begin{bmatrix}
\mu_p\\
1
\end{bmatrix}
=P
$$

Por lo tanto

$$
R^Tw=P
$$

Ahora juntamos los multiplicadores

$$
q=
\begin{bmatrix}
\gamma\\
\lambda
\end{bmatrix}
$$

Entonces

$$
Rq=
\begin{bmatrix}
\mu & \vec{1}
\end{bmatrix}
\begin{bmatrix}
\gamma\\
\lambda
\end{bmatrix}
$$

$$
Rq=\gamma\mu+\lambda\vec{1}
$$

Por lo que

$$
2\Sigma w+\lambda\vec{1}+\gamma\mu=0
$$

se puede escribir como

$$
2\Sigma w+Rq=0
$$

Despejamos $w$

$$
2\Sigma w=-Rq
$$

$$
\Sigma w=-\frac{1}{2}Rq
$$

Multiplicamos por $\Sigma^{-1}$

$$
\Sigma^{-1}\Sigma w=
-\frac{1}{2}\Sigma^{-1}Rq
$$

Como

$$
\Sigma^{-1}\Sigma=I
$$

queda

$$
Iw=-\frac{1}{2}\Sigma^{-1}Rq
$$

$$
w=-\frac{1}{2}\Sigma^{-1}Rq
$$

Sustituimos en la restricción

$$
R^Tw=P
$$

$$
R^T
\left(
-\frac{1}{2}\Sigma^{-1}Rq
\right)
=P
$$

$$
-\frac{1}{2}R^T\Sigma^{-1}Rq=P
$$

Multiplicamos por $-2$

$$
R^T\Sigma^{-1}Rq=-2P
$$

Ahora despejamos $q$

$$
\left(R^T\Sigma^{-1}R\right)^{-1}
\left(R^T\Sigma^{-1}R\right)q
=
-2
\left(R^T\Sigma^{-1}R\right)^{-1}P
$$

$$
Iq=
-2
\left(R^T\Sigma^{-1}R\right)^{-1}P
$$

$$
q=
-2
\left(R^T\Sigma^{-1}R\right)^{-1}P
$$

Sustituimos $q$ en la ecuación de $w$

$$
w=-\frac{1}{2}\Sigma^{-1}Rq
$$

$$
w=
-\frac{1}{2}\Sigma^{-1}R
\left[
-2
\left(R^T\Sigma^{-1}R\right)^{-1}P
\right]
$$

Como

$$
-\frac{1}{2}(-2)=1
$$

finalmente queda

$$
\boxed{
w^*_{\text{frontier}}
=
\Sigma^{-1}R
\left[
R^T\Sigma^{-1}R
\right]^{-1}P
}
$$


**Punto 2. Comprobación de Resultados Min Var y Max Sharpe (30%)**

+ Descarga el archivo de `data.xlsx` que subí a CANVAS, el cual contiene el histórico de los precios de cierre ajustado para `NVDA` y `AMZN`. Calcula los rendimientos diarios, utilizando el cambio porcentual. A posteriori, calcula el rendimiento esperado y la matriz de covarianza-varianza de los rendimientos diarios. Durante el ejercicio asume una tasa libre de riesgo ($r_f$) del $5\%$.


+ Con la información obtenida, calcula los pesos eficientes en mínima varianza y máximo de sharpe con la solución analítica obtenida en clase:

$$w_{min var} = \frac{\Sigma^{-1} \overrightarrow{1}}{\overrightarrow{1^T} \Sigma^{-1} \overrightarrow{1}}$$

$$w_{max sharpe} = \frac{\Sigma^{-1} (\mu - rf )}{\overrightarrow{1^T} \Sigma^{-1} (\mu - rf )}$$

+ Programa dos funciones para optimizar el portafolio en mínima varianza y máximo de sharpe utilizando la librería de `scipy.optimize.minimize`, utiliza las funciones para encontrar los pesos eficientes para el portafolio dado, deberás obtener los mismos resultados que con la solución analítica.

```python
# Importación de librerías
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize
```

```python
# Obtener precios
prices=pd.read_excel('data.xlsx')
prices=prices.set_index(['Date'])
prices
```

```python
# Obtener rendimiento
daily_rets = prices.pct_change().dropna()
daily_rets.head(2)
```

```python
# Obtener rendimiento promedio
mean = daily_rets.mean()
mean
```

```python
# Obtener matriz de covariânza varianza
cov = daily_rets.cov()
cov
```

```python
std = daily_rets.std()
std
```

```python
# Tasa libre de riesgo
rf=0.05/252
rf
```

```python
# Solución analítica Min Var
inv_cov=np.linalg.inv(cov)
ones=np.ones(len(prices.columns))

w_min_var_analytical=(inv_cov @ ones)/(ones.T @ inv_cov @ ones)
dict(zip(prices.columns,w_min_var_analytical))
```

```python
# Solución analítica Max Sharpe
w_max_sharpe_analytical=(inv_cov @ (mean-rf))/(ones.T @ inv_cov @ (mean-rf))
dict(zip(prices.columns,w_max_sharpe_analytical))
```

```python
# Programar funcion
def opt_min_var(prices: pd.DataFrame):
    #Calcular los rendimientos
    rets = prices.pct_change().dropna()

    #Calcular la covarianza: 
    cov=rets.cov()

    # Funcion. 
    var = lambda w: w.T @ cov @ w 

    #Variables iniciales (pesos)
    n_assets = len(prices.columns)

    w_0 = np.ones(n_assets) / n_assets 

    # Restricciones. 
    restr = lambda w: sum(w) - 1

    #rangos de las variables de decisión
    rangos=[(0,1)] * n_assets

    # tolerancia 
    tol=1e-32
    #Optimizar, port. eficiente min var
    results = minimize(fun=var, x0=w_0, bounds=rangos, constraints={'fun': restr, 'type': 'eq'}, tol=tol)
    return results.x #si regresa pesos equitativos no se hizo la optimización.
```

```python
# Verificar resultados de la función
w_min_var=opt_min_var(prices)
dict(zip(prices.columns,w_min_var))
```

```python
def opt_max_sharpe(price: pd.DataFrame, rf: float):

    # Inputs
    rets = price.pct_change().dropna()
    cov = rets.cov()
    mu = rets.mean()

    # Funcion
    sharpe = lambda w: -((sum(w*mu)-rf)/np.sqrt(w.T@cov@w))

    # Restricciones
    restr = lambda w: sum(w) - 1

    # Valor Inicial
    n_assets = len(price.columns)
    w_0 = np.ones(n_assets) / n_assets

    # tol
    tol = 1e-32

    # Rangos
    rangos = [(0,1)] * n_assets

    # optimización
    results = minimize(fun=sharpe,x0=w_0,bounds=rangos,constraints={'fun': restr, 'type': 'eq'},tol=tol)

    return results.x
```

```python
# Obtener los pesos
w_max_sharpe=opt_max_sharpe(prices, 0.05/252)
dict(zip(prices.columns, w_max_sharpe))
```

```python
# Comparar resultados
comparacion=pd.DataFrame({
    'Min Var Analítico':w_min_var_analytical*100,
    'Min Var Scipy':w_min_var*100,
    'Max Sharpe Analítico':w_max_sharpe_analytical*100,
    'Max Sharpe Scipy':w_max_sharpe*100
},index=prices.columns)

comparacion
```

Los resultados analíticos y los de scipy son prácticamente iguales en los dos portafolios.

En mínima varianza los pesos quedan aprox. 33.74% en NVIDIA y 66.26% en AMAZON.

En máximo Sharpe cambia la distribución y queda aprox. 71.71% en NVIDIA y 28.29% en AMAZON. Aquí se le da mucho más peso a NVIDIA porque estamos buscando una mejor relación entre rendimiento y riesgo y no solamente bajar la varianza.

**Punto 3. Gráfica de Frontera Eficiente en Media Varianza (40%)**

Con los dos puntos anteriores, gráfica la frontera eficiente analítica para el portafolio dado, la gráfica deberá contener los siguientes elementos:

+ Hipérbole de Markowitz (*"Markowitz Bullet"*)  


+ Coordenadas de los activos individuales según su volatilidad y rendimiento.


+ Coordenadas de los portafolios eficientes en mínima varianza y máximo de sharpe


+ Hipérbole de la Frontera Eficiente de Portafolios.


A continuación se muestra el producto esperado.





```python
# Frontera eficiente analítica
n_assets=len(prices.columns)
R=np.column_stack((mean,ones))
inv_R=np.linalg.inv(R.T @ inv_cov @ R)

mu_p=np.linspace(mean.min()-0.001,mean.max()+0.001,300)

rend_frontier=[]
vol_frontier=[]

for i in mu_p:
    P=np.array([i,1])
    w=inv_cov @ R @ inv_R @ P

    rend_frontier.append(sum(w*mean))
    vol_frontier.append(np.sqrt(w.T @ cov @ w))
```

```python
# Obtener rendimiento y volatilidad de Minima Varianza
R_min_var=sum(w_min_var_analytical*mean)
vol_min_var=np.sqrt(w_min_var_analytical.T @ cov @ w_min_var_analytical)

# Obtener rendimiento y volatilidad de Maximo de Sharpe
R_max_sharpe=sum(w_max_sharpe_analytical*mean)
vol_max_sharpe=np.sqrt(w_max_sharpe_analytical.T @ cov @ w_max_sharpe_analytical)

R_min_var,vol_min_var,R_max_sharpe,vol_max_sharpe
```

```python
# Realizar grafica
plt.figure(figsize=(12,6))

# Bala de Markowitz
plt.plot(vol_frontier,rend_frontier,label='Markowitz Bullet')

# Frontera Eficiente
rend_frontier=np.array(rend_frontier)
vol_frontier=np.array(vol_frontier)
eficiente=rend_frontier>=R_min_var
plt.plot(vol_frontier[eficiente],rend_frontier[eficiente],linewidth=3,label='Frontera Eficiente')

# Scatter plot de Minima Varianza
plt.scatter(vol_min_var,R_min_var,marker='X',s=100)
plt.text(vol_min_var,R_min_var,'Min Var')

# Scatter plot de Maximo de Sharpe
plt.scatter(vol_max_sharpe,R_max_sharpe,marker='X',s=100)
plt.text(vol_max_sharpe,R_max_sharpe,'Max Sharpe')

# Graficar coordenadas de activos individuales
for i in range(n_assets):
    plt.scatter(std.iloc[i],mean.iloc[i],marker='X',s=100)
    plt.text(std.iloc[i],mean.iloc[i],prices.columns[i])

# Nombrar eje x, y, titulo
plt.xlabel("Volatilidad")
plt.ylabel("Rendimiento")
plt.title('Bala de Markowitz y Frontera Eficiente en Media Varianza')
plt.legend()

plt.show()
```

En la gráfica la parte de arriba de la bala es la frontera eficiente porque ahí están las combinaciones que para cierto nivel de riesgo tienen el mayor rendimiento posible.

El portafolio de mínima varianza queda en la parte de la izquierda porque es el punto con menor riesgo. El de máximo Sharpe queda más arriba y a la derecha porque acepta más volatilidad pero también tiene un rendimiento esperado mayor.

También se pueden ver NVIDIA y AMAZON por separado para comparar sus coordenadas con los portafolios que salen de la optimización.
