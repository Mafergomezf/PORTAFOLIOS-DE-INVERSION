```python
import pandas as pd
import numpy as np
import yfinance as yf
```

```python
prices = yf.download(['WMT','NVDA'], start='2020-01-01', end='2026-09-07')['Close']

             
```

```python
rets=(prices/prices.shift()-1).dropna()
rets
```

```python
#Rend esperado
# la diferencia de usar la media o el CAGR es que la media no considera la capitalización de los rendimientos, mientras que el CAGR sí lo hace. Por lo tanto, el CAGR es una medida más precisa del rendimiento real de una inversión a lo largo del tiempo.
# Como te iria en promedio a futuro se usa la media 
rets.mean()*252*100
```

```python
#Media Geometrica CAGR

#CAGR<M  el CAGR se utiliza cuanod quieres ver el rendimiento real de una inversión a lo largo del tiempo, 
#considerando la capitalización de los rendimientos. Se calcula utilizando la fórmula:
#CAGR=
(((rets+1).prod())**(1/len(rets))-1)*252*100
```

```python
# hay dos tipos de medidas de rendimiento si hoy 7 de sept es hoy , 
# hay medidas de rendimiento que te sirven para el futuro forward looking que te dicen como te iria en promedio a futuro 
# y hay medidas de rendimiento que te dicen como te fue en el pasado backward looking que te dicen como te fue en el pasado.(real)

# el rendimeinto promedio diario es un forward looking y el CAGR es un backward looking.
```

```python
#Rend Armonico
(len(rets)/((1/(rets+1)).sum())-1)*252*100
```

## Rendimiento Esperado

```python
## Rendimiento Esperado
df = pd.DataFrame()
df['Escenario'] = ['Bajista', 'Neutral', 'Alcista']
df['Probabilidad'] = [0.25, 0.30, 0.45]
df['Rendimiento'] = [-10, 5, 20]

df
             
```

```python
sum(df['Probabilidad']*df['Rendimiento'])
#al hacer esta operación se obtiene el rendimiento esperado de la inversión, 
# considerando los diferentes escenarios y sus probabilidades asociadas 
```

```python
E_R=sum(df['Probabilidad']*df['Rendimiento'])
E_V=(df['Probabilidad']*(df['Rendimiento']-E_R)**2).sum()
E_vol=np.sqrt(E_V)
```

```python
E_R,E_V, E_vol
```
