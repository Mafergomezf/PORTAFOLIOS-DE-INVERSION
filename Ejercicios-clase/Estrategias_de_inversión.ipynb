## Estrategia de Inversión Pasiva

```python
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
import numpy as np
```

```python
prices=yf.download(['COHR','KO','SNOW','MRK','^GSPC'],start='2010-01-01', end='2026-09-01')['Close']
rets=prices.pct_change().dropna()
rets
```

```python
#Pasiva 
rets_pasiva=rets[['^GSPC']]

rf=0.05 #tasa libre de riesgo
E_R_pasiva=rets_pasiva.mean()*252
desvest_pasiva=rets_pasiva.std()*np.sqrt(252) #volatidad anualizada 

Sharpe_Ratio=(E_R_pasiva-rf)/desvest_pasiva 
#por cada unidad de riesgo asumida obtengo 0.67 unidades de rendimiento (en exceso a la tasa de rf)
Sharpe_Ratio
```

```python
#Activa
rets_activa=rets.drop(['^GSPC'],axis=1)
w=np.array([0.25,0.25,0.25,0.25]) #ponderaciones de cada activo
mean_activa=rets_activa.mean()*252
E_R_activa=sum(mean_activa*w)

cov=rets_activa.cov()
vol_R_activa=np.sqrt(w.T @ cov @ w)* np.sqrt(252)

sharpe_Ratio_activa=(E_R_activa-rf)/vol_R_activa
E_R_activa, vol_R_activa, sharpe_Ratio_activa
```
