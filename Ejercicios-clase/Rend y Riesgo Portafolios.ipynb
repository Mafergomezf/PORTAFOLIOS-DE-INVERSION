```python
import pandas as pd
import numpy as np
import yfinance as yf
```

```python
tickers=['NVDA','WMT']
```

```python
prices = yf.download(tickers, start='2024-01-01', end='2026-08-26')['Close']
prices
```

```python
#rend diarios
rets=(prices/prices.shift()-1).dropna()
rets
```

```python
#Rend promedio (Valor Esperado- Rendimiento Esperado)
mu=rets.mean()*252
mu*100
```

```python
#Rendimiento del Portafolio
w_port=np.array([0.5,0.5])

r_port=sum(w_port*mu)
r_port

#es la suma ponderada del peso que tiene el activo en el portafolio por su rendimiento
```

1. La covarianza mide la relacion lineal entre dos activos.   
2. No puede interpretar el valor o magnitud solo la dirección.    
3. La covarianza esta entre - infinito -- + infinito
4. Covarianza positiva me dice que la relación entre el activo x y y es positiva.   (si uno sube el otro tmb sube pero no se sabe cto)
5. Covarianza 0 indica que no hay relación lineal.   
6. Covarianza <0 la relación es negativa 

```python
#Riesgo portafolio
variance=rets.var()
```

```python
cov=rets.cov()
```

```python
varianza_port=w_port[0]**2*variance[0]+2*w_port[0]*w_port[1]*cov.iloc[0,1]+w_port[1]**2*variance[1]
desvest_port=np.sqrt(varianza_port)
varianza_port,desvest_port
```

```python
np.sqrt(w_port[0]*variance[0])
```

```python
w_port.T @ cov @ w_port,np.sqrt(w_port.T @ cov @ w_port)
```
