```python
import pandas as pd
import numpy as np
import yfinance as yf
import matplotlib.pyplot as plt
```

```python
tickers=['JPM','V','GS','PGR','^SP500-40']
prices=yf.download(tickers,start='2018-01-01',end='2026-09-07')['Close']
rets=(prices/prices.shift()-1)
rets
rf=0.04
```

## 1. ESTRATEGIA ACTIVA

```python
retsA=rets.drop(['^SP500-40'],axis=1).dropna()
w_portA=np.array([.25,.25,.25,.25])
cov=retsA.cov()
```

```python
daily_retsA=retsA.mean()
mu=daily_retsA*252

E_R_portA=sum(mu*w_portA)
E_R_portA*100
```

```python

daily_varianceA=w_portA.T@cov@w_portA
annual_varianceA=daily_varianceA*252
annual_varianceA
```

```python
daily_volatilityA=np.sqrt(daily_varianceA)
annual_volatilityA=np.sqrt(annual_varianceA)

annual_volatilityA*100
```

```python
Sharpe_RatioA= (E_R_portA-rf)/annual_volatilityA
Sharpe_RatioA
```

```python
cvA= annual_volatilityA/E_R_portA
cvA
```

```python
corr=retsA.corr()
corr
```

La correlación más alta se encuentra entre GS y JPM con 0.81, por lo que estos activos tienden a moverse de manera similar y aportan menos diversificación entre sí. La correlación más baja es entre GS y PGR con 0.32, por lo que estos tienen una relación menor y ayudan más a diversificar el portafolio.

```python
rets_portA = retsA @ w_portA

geometric_returnA = (((rets_portA+1).prod())**(1/len(rets_portA))-1)*252
geometric_returnA*100
```

```python
harmonic_returnA = (len(rets_portA)/((1/(rets_portA+1)).sum())-1)*252
harmonic_returnA*100
```

## 2. ESTRATEGIA PASIVA

Para la estrategia pasiva usé como benchmark el S&P 500 Financials con ticker ^SP500-40, porque representa al sector financiero y sirve como comparación para mi portafolio activo. Este índice incluye empresas financieras importantes del S&P 500, como JPMorgan Chase, Visa, Goldman Sachs, Bank of America, Mastercard, Berkshire Hathaway y Wells Fargo.

```python
retsP=rets[['^SP500-40']].dropna()
```

```python
daily_retsP=retsP.mean()
muP=daily_retsP*252
muP*100
```

```python
daily_varianceP=retsP.var()
annual_varianceP=daily_varianceP*252
annual_varianceP
```

```python
daily_volatilityP=np.sqrt(daily_varianceP)
annual_volatilityP=np.sqrt(annual_varianceP)
annual_volatilityP*100
```

```python
Sharpe_RatioP= (muP-rf)/annual_volatilityP
Sharpe_RatioP
```

```python
cvP= annual_volatilityP/muP
cvP
```

```python
geometric_returnP=(((retsP+1).prod())**(1/len(retsP))-1)*252
geometric_returnP*100
```

```python
harmonic_returnP=(len(retsP)/((1/(retsP+1)).sum())-1)*252
harmonic_returnP*100
```
