```python
import pandas as pd
import numpy as np
import yfinance as yf
```

```python
tickers = ['NVDA', 'WMT']
```

```python
prices = yf.download(tickers, start='2024-01-01', end='2026-08-26')['Close']
prices
```

```python
# Calcular los rendimientos
Returns = (prices/prices.shift() - 1).dropna() # Eliminar las filas con valores NaN
Returns
```

```python
# Calcular Valor Esperado (Expected Value)
expected_returns = Returns.mean()  # Asumiendo que los rendimientos son independientes y distribuidos idénticamente
print(expected_returns * 100)  # Convertir a porcentaje
```

```python
# Rendmiento del Portafolio
W = np.array([0.5, 0.5])  # Pesos del portafolio (50% NVDA, 50% WMT)
portfolio_return = sum(expected_returns * W) # Calcular el rendimiento del portafolio
print(portfolio_return * 100)  # Convertir a porcentaje
```

```python
# Riesgo Portafolio
Variance = Returns.var()  # Matriz de covarianza diaria
Covariance = Returns.cov().iloc[0, 1] # Matriz de covarianza anualizada
```

```python
# Calcular la varianza del portafolio
Varianza_Portafolio = W[0]**2 * Variance[0] + W[1]**2 * Variance[1] + 2 * W[0] * W[1] * Covariance
Desviacion_Portafolio = np.sqrt(Varianza_Portafolio)

print("Varianza del Portafolio:", Varianza_Portafolio)
print("Desviación del Portafolio:", Desviacion_Portafolio)
```
