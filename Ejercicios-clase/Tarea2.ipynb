## Selección de activos

Para formar el portafolio se seleccionaron cuatro activos buscando que pertenezcan a mercados o tipos de exposición diferentes, con el objetivo de reducir la correlación entre ellos y mejorar la diversificación.

Los activos seleccionados tendrán una ponderación igual dentro del portafolio.

```python
#Definir activos
tickers=['SPY', 'GLD', 'DBA', 'UUP']
```

```python
#descargar datos
prices=yf.download(tickers,start='2020-01-01',end='2026-09-01')["Close"]
prices
```

```python
#rend diarios
rets=(prices/prices.shift()-1).dropna()
rets
```

```python
#rend promedio
mu=rets.mean()*252
mu*100
```

```python
#Rendimiento del Portafolio
w_port=np.array([0.25,0.25,0.25,0.25])

r_port=sum(w_port*mu)
r_port

#es la suma ponderada del peso que tiene el activo en el portafolio por su rendimiento
```
