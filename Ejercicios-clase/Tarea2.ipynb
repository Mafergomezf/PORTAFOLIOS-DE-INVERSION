```python
#importar librerias
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import yfinance as yf
```

```python
#Definir activos
tickers=['SPY', 'GLD', 'DBA', 'BTC-USD']
```

```python
#descargar datos
prices=yf.download(tickers,start='2020-01-01',end='2026-09-01')["Close"]
prices
```

```python
#rend diarios
rets=(prices/prices.shift()-1).dropna()
rets
```

```python
#Rend promedio (Valor Esperado- Rendimiento Esperado)
mu=rets.mean()*252
mu*100
```

```python
#Rendimiento del Portafolio
w_port=np.array([0.25,0.25,0.25,0.25])

r_port=sum(w_port*mu)
r_port

#es la suma ponderada del peso que tiene el activo en el portafolio por su rendimiento
```

```python
#Riesgo portafolio
variance=rets.var()
```

```python
cov=rets.cov()
```

```python
w_port.T @ cov @ w_port,np.sqrt(w_port.T @ cov @ w_port)
```
