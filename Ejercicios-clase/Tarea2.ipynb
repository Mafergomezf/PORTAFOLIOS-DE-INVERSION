```python
#importar librerias
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import yfinance as yf
```

```python
#Definir activos
tickers=['SPY', 'GLD', 'DBA', 'UUP']
```

```python
#descargar datos
prices=yf.download(tickers,start='2020-01-01',end='2026-09-01')["Close"]
prices
```

```python
#rend diarios
rets=(prices/prices.shift()-1).dropna()
rets
```

## Rendimiento diario y anual del portafolio

```python
#Rendimiento esperado de cada activo
mu=rets.mean()
```

```python
#peso de los activos en el portafolio
w_port=np.array([0.25,0.25,0.25,0.25])
```

```python
#Rendimiento del Portafolio
r_port=sum(w_port*mu)
r_port
```

```python
#rend promedio
mu=rets.mean()*252
mu*100
```

```python
#Rendimiento del Portafolio
r_port=sum(w_port*mu)
r_port

#es la suma ponderada del peso que tiene el activo en el portafolio por su rendimiento
```

## Varianza diaria y anual del portafolio

```python
#Riesgo portafolio
variance=rets.var()
variance
```

```python
cov=rets.cov()
cov
```

```python
daily_variance=w_port.T @ cov @ w_port
```

```python
daily_variance
```

```python
annual_variance=daily_variance*252
annual_variance*100
```

## Volatilidad diaria y anual del portafolio

```python
daily_volatility=np.sqrt(w_port.T @ cov @ w_port)
```

```python
daily_volatility*100
```

```python
volatility_annual=daily_volatility*np.sqrt(252)
volatility_annual*100
```

## Matriz de correlación

```python
#Matriz de correlación
rets.corr()
```

## Conclusión sobre la matriz de correlación y el riesgo no sistemático de cada activo

### Matriz de correlación

La matriz muestra que los activos presentan correlaciones relativamente bajas entre sí. DBA, GLD y SPY tienen relaciones positivas pero débiles, mientras que UUP presenta correlaciones negativas con los otros tres activos.

La correlación más negativa se observa entre GLD y UUP, con un valor aproximado de -0.43, lo que indica que tienden a moverse en direcciones opuestas en cierta medida.

En general, la combinación de estos activos favorece la diversificación del portafolio, ya que no se encuentran altamente correlacionados entre sí.

### Riesgo no sistemático

Al comparar las varianzas individuales, SPY presenta el mayor nivel de riesgo entre los cuatro activos, seguido de GLD y DBA, mientras que UUP presenta la menor varianza.

Sin embargo, el riesgo del portafolio no depende solamente del riesgo individual de cada activo, sino también de la relación entre ellos. Como las correlaciones son en general bajas y algunas son negativas, la combinación de los cuatro activos permite reducir parte del riesgo no sistemático.

En especial, UUP ayuda a diversificar el portafolio porque presenta correlaciones negativas con DBA, GLD y SPY.
