```python
#importar librerias
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import yfinance as yf
```

## Selección de activos

Se seleccionaron los activos BDRY, FXY, WEAT y XLP.
- BDRY representa fletes marítimos.
- FXY representa el yen japonés.
- WEAT representa el mercado del trigo.
- XLP representa empresas de consumo básico.

Se escogieron porque pertenecen a mercados distintos, lo que ayuda a que el portafolio tenga una mejor diversificación.

```python
#Definir activos
tickers=['BDRY','FXY','WEAT','XLP']
```

```python
#descargar datos
prices=yf.download(tickers,start='2020-01-01',end='2026-09-01')["Close"]
prices
```

```python
#rend diarios
rets=(prices/prices.shift()-1).dropna()
rets
```

## Rendimiento diario y anual del portafolio

```python
#Rendimiento esperado de cada activo
mu=rets.mean()
```

```python
#peso de los activos en el portafolio
w_port=np.array([0.25,0.25,0.25,0.25])
```

```python
#Rendimiento del Portafolio
r_port=sum(w_port*mu)
r_port
```

```python
#rend promedio
mu=rets.mean()*252
mu*100
```

```python
#Rendimiento del Portafolio
r_port=sum(w_port*mu)
r_port

#es la suma ponderada del peso que tiene el activo en el portafolio por su rendimiento
```

## Varianza diaria y anual del portafolio

```python
#Riesgo portafolio
variance=rets.var()
variance
```

```python
cov=rets.cov()
cov
```

```python
daily_variance=w_port.T @ cov @ w_port
```

```python
daily_variance
```

```python
annual_variance=daily_variance*252
annual_variance*100
```

## Volatilidad diaria y anual del portafolio

```python
daily_volatility=np.sqrt(w_port.T @ cov @ w_port)
```

```python
daily_volatility*100
```

```python
volatility_annual=daily_volatility*np.sqrt(252)
volatility_annual*100
```

## Matriz de correlación

```python
#Matriz de correlación
rets.corr()
```

### Conclusión sobre la matriz de correlación y el riesgo no sistemático

Lo primero que nos llamó la atención de la matriz es que practicamente todas las correlaciones quedaron muy cercanas a 0, entonces los cuatro activos casi no se movieron de la misma manera durante el periodo que analizamos. Por ejemplo BDRY y XLP tuvieron una correlación de solamente 0.0053 y FXY con XLP de -0.0011, incluso la correlación mas alta de toda la matriz fue FXY con WEAT y fue de apenas 0.0399, entonces realmente sigue siendo bastante baja.

Esto tiene sentido por los activos que escogimos porque aunque todos forman parte del mismo portafolio, cada uno depende de cosas muy diferentes. BDRY esta relacionado con los fletes marítimos, FXY con el yen japonés, WEAT con el trigo y XLP con empresas de consumo básico. Entonces si por ejemplo pasa algo que afecte especificamente al mercado del trigo no necesariamente tendría que afectar de la misma forma al yen o a las empresas de consumo básico.

En cuanto al riesgo no sistemático, cada activo tambien tiene riesgos propios. BDRY puede verse afectado por cambios en los fletes marítimos y en la demanda de transporte, FXY por movimientos propios del yen japonés, WEAT por factores como la producción de trigo, el clima o cambios en la oferta agrícola y XLP por problemas específicos de las empresas del sector de consumo básico. En nuestros resultados BDRY fue el activo con mayor varianza y FXY el de menor, mientras que WEAT y XLP quedaron en medio.

Esto no quiere decir que el riesgo especifico de cada activo desaparezca al meterlo al portafolio, mas bien al combinarlos con activos que casi no estan correlacionados entre si se puede disminuir el efecto que tendría un problema de uno solo sobre todo el portafolio. Por eso creemos que en este caso la diversificación si ayuda bastante a reducir el riesgo no sistemático, porque los riesgos vienen de fuentes diferentes y los movimientos de los cuatro activos casi no se repiten.

### Conclusión sobre las métricas de rendimiento, riesgo y diversificación del portafolio

Para nosotros lo mas importante de este portafolio fue la diversificación, porque las correlaciones entre los cuatro activos quedaron practicamente en 0. Esto quiere decir que no se estuvieron moviendo todos de la misma manera y entonces el portafolio no depende tanto de lo que pase en un solo mercado. Por ejemplo si algo afecta al trigo no necesariamente va a afectar igual al yen, a los fletes marítimos o a las empresas de consumo básico.

Entonces creemos que el portafolio si cumple bien con lo que buscábamos desde el inicio, porque obtuvimos un rendimiento positivo pero además logramos combinar activos que casi no estan relacionados entre ellos. No es que por estar diversificado ya no exista riesgo, mas bien la idea es que el riesgo especifico de un activo no afecte de la misma forma a todo el portafolio y por eso creemos que la diversificación que obtuvimos fue bastante buena.
